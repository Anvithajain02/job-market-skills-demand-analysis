```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("/content/indian_tech_jobs_2026.csv")

print("Dataset loaded successfully!")
print("Rows and columns:", df.shape)

display(df.head())
```

```python

# STEP 3: UNDERSTANDING THE DATASET

# 1. Display the number of rows and columns
print("Dataset shape:", df.shape)

# 2. Display all column names
print("\nColumn names:")
print(df.columns.tolist())

# 3. Display the first 5 records
print("\nFirst 5 records:")
display(df.head())

# 4. Display the last 5 records
print("\nLast 5 records:")
display(df.tail())

# 5. Check data types and missing values
print("\nDataset information:")
df.info()

# 6. Count missing values in each column
print("\nMissing values:")
print(df.isnull().sum())

# 7. Count duplicate rows
print("\nDuplicate rows:", df.duplicated().sum())
```

```python

# STEP 4: DATA CLEANING

# Make a copy so we preserve the original dataset
df_clean = df.copy()

# Check missing values
print("Missing values in each column:")
display(
    df_clean.isnull().sum()
    .sort_values(ascending=False)
    .to_frame("Missing Values")
)

# Check exact duplicate rows
print("Duplicate rows:", df_clean.duplicated().sum())

# Check duplicate job IDs
print("Duplicate job IDs:", df_clean["job_id"].duplicated().sum())

# Check blank job titles
print(
    "Blank job titles:",
    df_clean["job_title"].astype(str).str.strip().eq("").sum()
)

print("\nOriginal rows:", len(df))
print("Working rows:", len(df_clean))
```

```python

# Convert numeric columns to appropriate types
numeric_columns = [
    "company_rating",
    "experience_min_yrs",
    "experience_max_yrs",
    "salary_min_lpa",
    "salary_max_lpa",
    "salary_midpoint_lpa",
    "skills_count",
    "days_since_posted"
]

for column in numeric_columns:
    df_clean[column] = pd.to_numeric(
        df_clean[column],
        errors="coerce"
    )

# Convert boolean columns
boolean_columns = [
    "salary_disclosed",
    "is_senior",
    "is_fresher_friendly",
    "salary_negotiable"
]

for column in boolean_columns:
    df_clean[column] = df_clean[column].astype("boolean")

# Convert date fields
df_clean["scraped_at"] = pd.to_datetime(
    df_clean["scraped_at"],
    errors="coerce"
)

# Standardize text spacing
text_columns = [
    "job_title",
    "company_name",
    "location",
    "role_category",
    "skills_required",
    "work_mode",
    "primary_city",
    "skill_domain"
]

for column in text_columns:
    df_clean[column] = (
        df_clean[column]
        .astype("string")
        .str.strip()
    )

print("Data types updated.")
display(df_clean.dtypes.to_frame("Data Type"))
```

```python

# Inspect salary information
print("Salary disclosure counts:")
display(df_clean["salary_disclosed"].value_counts(dropna=False))

print("\nSalary ranges:")
display(
    df_clean[
        ["salary_min_lpa", "salary_max_lpa", "salary_midpoint_lpa"]
    ].describe().round(2)
)

print("\nExperience ranges:")
display(
    df_clean[
        ["experience_min_yrs", "experience_max_yrs"]
    ].describe().round(2)
)

print("\nWork modes:")
display(df_clean["work_mode"].value_counts(dropna=False))

print("\nJob role categories:")
display(df_clean["role_category"].value_counts().head(15))
```

```python

# STEP 5.1: MOST COMMON JOB ROLES

role_counts = (
    df_clean["role_category"]
    .value_counts()
    .reset_index()
)

role_counts.columns = ["Job Role", "Number of Postings"]

# Calculate percentage of all postings
role_counts["Percentage"] = (
    role_counts["Number of Postings"]
    / len(df_clean) * 100
).round(2)

display(role_counts)

# Visualize top 10 roles
plt.figure(figsize=(11, 6))

sns.barplot(
    data=role_counts.head(10),
    x="Number of Postings",
    y="Job Role",
    hue="Job Role",
    legend=False
)

plt.title("Top 10 Job Roles in the Dataset")
plt.xlabel("Number of Job Postings")
plt.ylabel("Job Role")
plt.tight_layout()
plt.show()
```

```python

# STEP 5.2: INSPECT THE SKILLS COLUMN

print("Examples of required skills:")

display(
    df_clean[
        ["job_title", "role_category", "skills_required"]
    ].head(10)
)
```

```python

# STEP 5.3: SALARY ANALYSIS
# Use only postings where salary is disclosed.

salary_data = df_clean[
    (df_clean["salary_disclosed"] == True)
    & (df_clean["salary_min_lpa"] > 0)
    & (df_clean["salary_max_lpa"] >= df_clean["salary_min_lpa"])
].copy()

print("Total job postings:", len(df_clean))
print("Postings with disclosed salary:", len(salary_data))

if not salary_data.empty:
    print("\nSalary summary (LPA):")

    display(
        salary_data[
            ["salary_min_lpa", "salary_max_lpa", "salary_midpoint_lpa"]
        ].describe().round(2)
    )

    # Median salary midpoint by job role
    salary_by_role = (
        salary_data.groupby("role_category")
        .agg(
            Job_Postings=("job_id", "count"),
            Median_Salary_LPA=("salary_midpoint_lpa", "median"),
            Average_Salary_LPA=("salary_midpoint_lpa", "mean")
        )
        .sort_values("Median_Salary_LPA", ascending=False)
        .round(2)
        .reset_index()
    )

    display(salary_by_role)

    plt.figure(figsize=(11, 6))

    sns.barplot(
        data=salary_by_role,
        x="Median_Salary_LPA",
        y="role_category",
        hue="role_category",
        legend=False
    )

    plt.title("Median Disclosed Salary by Job Role")
    plt.xlabel("Median Salary (LPA)")
    plt.ylabel("Job Role")
    plt.tight_layout()
    plt.show()
else:
    print("No valid disclosed salary records found.")
```

```python

# STEP 6.1: Inspect required skills

display(
    df_clean[["job_title", "role_category", "skills_required"]].head(10)
)

print("\nExample skill values:")
for value in df_clean["skills_required"].head(5):
    print(value)
```

```python

# STEP 6.2: Clean and extract skills

import re
from collections import Counter

def extract_skills(skill_text):
    if pd.isna(skill_text):
        return []

    text = str(skill_text).strip()

    # Treat common representations of missing data as empty
    if text.lower() in ["", "nan", "none", "null", "not specified"]:
        return []

    # Split the skill list using common separators
    parts = re.split(r"[,;|\n]+", text)

    # Normalize whitespace and remove empty entries
    skills = [
        re.sub(r"\s+", " ", part).strip()
        for part in parts
    ]

    # Remove duplicate skills within each job posting
    return list(dict.fromkeys(
        skill for skill in skills if skill
    ))

df_clean["skills_list"] = df_clean["skills_required"].apply(extract_skills)

print("Skill extraction completed!")
display(
    df_clean[["job_title", "skills_required", "skills_list"]].head(10)
)
```

```python

# STEP 6.3: Count skills across job postings

skill_counter = Counter(
    skill
    for skills in df_clean["skills_list"]
    for skill in skills
)

skill_demand = pd.DataFrame(
    skill_counter.items(),
    columns=["Skill", "Job_Postings"]
)

# Count how many job postings mention each skill
skill_demand["Percentage"] = (
    skill_demand["Job_Postings"] / len(df_clean) * 100
).round(2)

skill_demand = skill_demand.sort_values(
    "Job_Postings", ascending=False
).reset_index(drop=True)

print("Total unique skill labels:", len(skill_demand))
print("\nTop 20 most demanded skills:")

display(skill_demand.head(20))
```

```python

# STEP 6.4: Visualize top 15 skills

top_skills = (
    skill_demand.head(15)
    .sort_values("Job_Postings", ascending=True)
)

plt.figure(figsize=(12, 7))

sns.barplot(
    data=top_skills,
    x="Job_Postings",
    y="Skill",
    hue="Skill",
    legend=False
)

plt.title("Top 15 Skills in the Job Market", fontsize=16)
plt.xlabel("Number of Job Postings")
plt.ylabel("Technical Skill")
plt.tight_layout()

plt.savefig(
    "top_15_skills.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python

# STEP 7: Compare skills across roles

role_skill_data = df_clean[
    ["role_category", "skills_list"]
].explode("skills_list")

role_skill_data = role_skill_data.dropna(
    subset=["skills_list"]
)

role_skill_counts = (
    role_skill_data.groupby(
        ["role_category", "skills_list"]
    )
    .size()
    .reset_index(name="Job_Postings")
    .rename(columns={"skills_list": "Skill"})
)

# Select the 15 most common skills overall
top_skill_names = skill_demand.head(15)["Skill"].tolist()

role_skill_top = role_skill_counts[
    role_skill_counts["Skill"].isin(top_skill_names)
]

role_skill_matrix = role_skill_top.pivot(
    index="Skill",
    columns="role_category",
    values="Job_Postings"
).fillna(0)

plt.figure(figsize=(14, 9))

sns.heatmap(
    role_skill_matrix,
    cmap="YlGnBu",
    annot=True,
    fmt=".0f",
    linewidths=0.3
)

plt.title("Skill Demand Across Job Roles")
plt.xlabel("Job Role")
plt.ylabel("Skill")
plt.tight_layout()

plt.savefig(
    "skills_by_job_role.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python

# STEP 8.1: NORMALIZE SKILL NAMES

def normalize_skill(skill):
    skill = str(skill).strip().lower()

    # Standardize common variations
    aliases = {
        "python": "Python",
        "sql": "SQL",
        "data analysis": "Data Analysis",
        "analytics": "Analytics",
        "analytical": "Analytical",
        "business analysis": "Business Analysis",
        "business analyst": "Business Analyst",
        "machine learning": "Machine Learning",
        "artificial intelligence": "Artificial Intelligence",
        "agile": "Agile"
    }

    return aliases.get(skill, skill.title())


df_clean["normalized_skills"] = df_clean["skills_list"].apply(
    lambda skills: list(dict.fromkeys(
        normalize_skill(skill) for skill in skills
    ))
)

print("Original skill examples:")
display(df_clean["skills_list"].head())

print("\nNormalized skill examples:")
display(df_clean["normalized_skills"].head())
```

```python

# STEP 8.2: RECALCULATE SKILL FREQUENCIES

from collections import Counter

normalized_counter = Counter(
    skill
    for skills in df_clean["normalized_skills"]
    for skill in set(skills)
)

skill_demand = pd.DataFrame(
    normalized_counter.items(),
    columns=["Skill", "Job_Postings"]
)

skill_demand["Percentage"] = (
    skill_demand["Job_Postings"] / len(df_clean) * 100
).round(2)

skill_demand = skill_demand.sort_values(
    "Job_Postings", ascending=False
).reset_index(drop=True)

display(skill_demand.head(20))
```

```python

# STEP 8.3: FIXED ROLE-WISE SKILLS HEATMAP

# Keep job_id so duplicate skills can be removed correctly
role_skill_data = df_clean[
    ["job_id", "role_category", "normalized_skills"]
].explode("normalized_skills")

# Remove rows without a detected skill
role_skill_data = role_skill_data.dropna(
    subset=["normalized_skills"]
)

# Remove duplicate skill entries within the same job
role_skill_data = role_skill_data.drop_duplicates(
    subset=["job_id", "role_category", "normalized_skills"]
)

# Count each skill for each job role
role_skill_counts = (
    role_skill_data.groupby(
        ["normalized_skills", "role_category"]
    )
    .size()
    .reset_index(name="Job_Postings")
)

# Create a matrix for the heatmap
role_skill_matrix = role_skill_counts.pivot(
    index="normalized_skills",
    columns="role_category",
    values="Job_Postings"
).fillna(0)

# Select the 15 most frequent skills
top_skill_names = skill_demand.head(15)["Skill"].tolist()

role_skill_matrix = role_skill_matrix.loc[
    role_skill_matrix.index.isin(top_skill_names)
]

# Plot the heatmap
plt.figure(figsize=(13, 8))

sns.heatmap(
    role_skill_matrix,
    annot=True,
    fmt=".0f",
    cmap="YlGnBu",
    linewidths=0.3
)

plt.title("Top Skills Across Job Roles")
plt.xlabel("Job Role")
plt.ylabel("Skill")
plt.xticks(rotation=35, ha="right")
plt.tight_layout()

plt.savefig(
    "cleaned_skills_by_job_role.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Heatmap generated successfully!")
```

```python

# STEP 9: FRESHER-FRIENDLY JOB ANALYSIS

fresher_jobs = df_clean[
    df_clean["is_fresher_friendly"] == True
].copy()

print("Total job postings:", len(df_clean))
print("Fresher-friendly postings:", len(fresher_jobs))

print(
    "Percentage of postings marked fresher-friendly:",
    round(len(fresher_jobs) / len(df_clean) * 100, 2),
    "%"
)

fresher_role_counts = (
    fresher_jobs["role_category"]
    .value_counts()
    .reset_index()
)

fresher_role_counts.columns = ["Job Role", "Job Postings"]

display(fresher_role_counts)

plt.figure(figsize=(11, 6))

sns.barplot(
    data=fresher_role_counts,
    x="Job Postings",
    y="Job Role",
    hue="Job Role",
    legend=False
)

plt.title("Fresher-Friendly Job Postings by Role")
plt.xlabel("Number of Postings")
plt.ylabel("Job Role")
plt.tight_layout()

plt.savefig(
    "fresher_friendly_jobs.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python

# STEP 10: JOB LOCATION ANALYSIS

# Count job postings by primary city
city_counts = (
    df_clean["primary_city"]
    .value_counts()
    .reset_index()
)

city_counts.columns = ["City", "Job_Postings"]

# Calculate percentage of all job postings
city_counts["Percentage"] = (
    city_counts["Job_Postings"] / len(df_clean) * 100
).round(2)

print("Top 15 cities by job posting count:")
display(city_counts.head(15))

# Visualize top 15 cities
top_cities = (
    city_counts.head(15)
    .sort_values("Job_Postings", ascending=True)
)

plt.figure(figsize=(11, 7))

sns.barplot(
    data=top_cities,
    x="Job_Postings",
    y="City",
    hue="City",
    legend=False
)

plt.title("Top 15 Cities by Job Postings", fontsize=15)
plt.xlabel("Number of Job Postings")
plt.ylabel("City")
plt.tight_layout()

plt.savefig(
    "top_15_job_cities.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python

# STEP 11: WORK MODE ANALYSIS

work_mode_counts = (
    df_clean["work_mode"]
    .value_counts()
    .reset_index()
)

work_mode_counts.columns = ["Work_Mode", "Job_Postings"]

work_mode_counts["Percentage"] = (
    work_mode_counts["Job_Postings"] / len(df_clean) * 100
).round(2)

display(work_mode_counts)

plt.figure(figsize=(9, 6))

sns.barplot(
    data=work_mode_counts,
    x="Work_Mode",
    y="Job_Postings",
    hue="Work_Mode",
    legend=False
)

plt.title("Job Postings by Work Mode", fontsize=15)
plt.xlabel("Work Mode")
plt.ylabel("Number of Job Postings")
plt.tight_layout()

plt.savefig(
    "job_work_mode_analysis.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python

# STEP 13: ROLE-WISE SKILL DEMAND PERCENTAGES

# Count each distinct skill once per job posting
role_skill_data = df_clean[
    ["job_id", "role_category", "normalized_skills"]
].explode("normalized_skills")

role_skill_data = role_skill_data.dropna(
    subset=["normalized_skills"]
).drop_duplicates(
    subset=["job_id", "normalized_skills"]
)

# Number of postings in each role
role_totals = df_clean.groupby("role_category")["job_id"].nunique()

# Count postings mentioning each skill in each role
role_skill_counts = (
    role_skill_data.groupby(["role_category", "normalized_skills"])
    ["job_id"]
    .nunique()
    .reset_index(name="Job_Postings")
)

role_skill_counts["Total_Role_Postings"] = (
    role_skill_counts["role_category"].map(role_totals)
)

role_skill_counts["Percentage"] = (
    role_skill_counts["Job_Postings"]
    / role_skill_counts["Total_Role_Postings"] * 100
).round(2)

print("Top skills by role, sorted by percentage:")
display(
    role_skill_counts.sort_values(
        ["role_category", "Percentage"],
        ascending=[True, False]
    ).groupby("role_category").head(10)
)

# Create a percentage heatmap using the 15 most frequent skills
top_skill_names = skill_demand.head(15)["Skill"].tolist()

percentage_matrix = role_skill_counts[
    role_skill_counts["normalized_skills"].isin(top_skill_names)
].pivot(
    index="normalized_skills",
    columns="role_category",
    values="Percentage"
).fillna(0)

plt.figure(figsize=(13, 8))
sns.heatmap(
    percentage_matrix,
    annot=True,
    fmt=".1f",
    cmap="YlGnBu"
)

plt.title("Percentage of Job Postings Mentioning Each Skill by Role")
plt.xlabel("Job Role")
plt.ylabel("Skill")
plt.tight_layout()
plt.savefig("role_skill_percentages.png", dpi=300, bbox_inches="tight")
plt.show()
```
